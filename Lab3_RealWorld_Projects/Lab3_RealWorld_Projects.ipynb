```python
LAST_NAME = "MEDINA"
SEED_NUM = 5
FAVORITE_ARTIST = "ERASERHEAD"


def generate_readings():
    base = sum(map(ord, LAST_NAME)) + sum(map(ord, FAVORITE_ARTIST)) + SEED_NUM

    readings = []

    for i in range(1, 7):
        if i == 5:
            readings.append(("ERR", "N/A", 1.20))
        else:
            temperature = round(30 + ((base + i * 3) % 31) / 10, 1)
            voltage = round(10 + ((base * 2 + i * 4) % 41) / 10, 1)
            current = round(0.50 + ((base + i * 5) % 31) / 100, 2)

            readings.append((temperature, voltage, current))

    return readings


def validate_reading(reading):
    temperature, voltage, current = reading

    if not all(isinstance(x, (int, float)) for x in reading):
        return False

    return (
        0 <= temperature <= 100
        and 0 <= voltage <= 240
        and 0 <= current <= 20
    )


def calculate_power(voltage, current):
    return round(voltage * current, 2)


def classify_condition(temperature, power):
    if temperature >= 70 or power >= 1000:
        return "CRITICAL"
    elif temperature >= 50 or power >= 500:
        return "WARNING"
    else:
        return "NORMAL"


def main():
    print("=== EQUIPMENT DIAGNOSTIC SYSTEM ===")

    print("Student:", LAST_NAME)
    print("Seed Number:", SEED_NUM)
    print("Favorite Artist:", FAVORITE_ARTIST)

    readings = generate_readings()

    print("\nGenerated Equipment Data:")

    for i, reading in enumerate(readings, 1):
        print(f"Reading {i}: {reading}")

    valid_results = []
    invalid_count = 0

    print("\nValidation Results:")

    for i, reading in enumerate(readings, 1):

        try:
            if not validate_reading(reading):
                raise ValueError("Invalid equipment reading")

            temperature, voltage, current = reading

            power = calculate_power(voltage, current)
            condition = classify_condition(temperature, power)

            valid_results.append(
                (i, temperature, voltage, current, power, condition)
            )

            print(f"Reading {i}: VALID")

        except (ValueError, TypeError) as error:
            invalid_count += 1
            print(f"Reading {i}: INVALID - {error}")

    print("\nDiagnostic Results:")

    for result in valid_results:
        i, temperature, voltage, current, power, condition = result

        print(
            f"Reading {i}: "
            f"Power = {power:.2f} W, "
            f"Condition = {condition}"
        )

    if any(result[-1] != "NORMAL" for result in valid_results):
        status = "ABNORMAL"
    else:
        status = "NORMAL"

    print("\nFinal Output:")
    print("Valid readings:", len(valid_results))
    print("Invalid readings:", invalid_count)
    print("Overall equipment status:", status)

    print(
        "Execution Log: "
        "Generation -> Validation -> Calculation -> "
        "Classification -> Summary"
    )


if __name__ == "__main__":
    main()
```

```python
LAST_NAME = "MEDINA"
SEED_NUM = 5
FAVORITE_ARTIST = "ERASERHEAD"


def generate_fault_code():
    last_name_value = sum(map(ord, LAST_NAME))
    artist_value = sum(map(ord, FAVORITE_ARTIST))

    fault_code = (
        last_name_value * SEED_NUM + artist_value
    ) % 10000

    return fault_code


def trace_fault(code, trace=None):

    if trace is None:
        trace = []

    trace.append(code)

    # Base condition
    if code <= 0:
        return trace

    return trace_fault(code // 2, trace)


def main():

    print("=== RECURSIVE FAULT TRACE ===")

    print("Student:", LAST_NAME)
    print("Seed Number:", SEED_NUM)
    print("Favorite Artist:", FAVORITE_ARTIST)

    fault_code = generate_fault_code()

    trace = trace_fault(fault_code)

    print("\nGenerated Fault Data:")
    print(fault_code)

    print("\nRecursive Trace:")
    print(" -> ".join(map(str, trace)))

    print("\nNumber of Recursive Calls:")
    print(len(trace))

    print("\nExecution Log:")
    print(
        "Generate Fault Code -> "
        "Recursive Trace -> "
        "Base Condition Reached"
    )

    print("\nFinal Output:")
    print(f"Fault {fault_code} successfully traced to 0.")


if __name__ == "__main__":
    main()
```

```python
def generate_telemetry(last_name, seed_num, artist):
    """
    Generator that produces telemetry readings one at a time.
    The complete telemetry stream is not stored in memory.
    """

    profiles = [
        (72, 2.1, 18),
        (75, 2.4, 20),
        (91, 3.8, 28),
        (68, 1.9, 17),
        (105, 4.7, 36),   # Invalid
        (77, 2.6, 22),
        (82, 3.1, 25),
        (69, 2.0, 19),
        (115, 5.2, 41),   # Invalid
        (74, 2.3, 21)
    ]

    base = sum(ord(c) for c in last_name) + seed_num * 11 + len(artist)

    for i, (temp, vibration, pressure) in enumerate(profiles, start=1):

        shift = ((base + i * seed_num) % 3) - 1

        yield {
            "id": i,
            "temperature": temp + shift,
            "vibration": round(vibration + shift * 0.1, 1),
            "pressure": pressure + shift,
            "source": f"{last_name}-{artist.upper()}-{seed_num}"
        }


def validate_reading(reading):
    """
    Validates whether the telemetry reading is within
    the acceptable operating range.
    """

    return (
        60 <= reading["temperature"] <= 100
        and 0 <= reading["vibration"] <= 4.0
        and 15 <= reading["pressure"] <= 35
    )

from functools import wraps


def monitor(func):
    """
    Decorator used to monitor the major processing function.
    """

    @wraps(func)
    def wrapper(*args, **kwargs):

        wrapper.calls += 1

        print(
            f"[DECORATOR] {func.__name__}() "
            f"call #{wrapper.calls}"
        )

        return func(*args, **kwargs)

    wrapper.calls = 0

    return wrapper


@monitor
def process_reading(reading):
    """
    Processes a valid telemetry reading and
    calculates an equipment health score.
    """

    score = (
        reading["temperature"] / 100
        + reading["vibration"] / 4
        + reading["pressure"] / 35
    ) / 3

    # Lambda transforms the score into a percentage.
    to_percentage = lambda x: round(x * 100, 2)

    health_score = to_percentage(score)

    if score >= 0.90:
        status = "CRITICAL"

    elif score >= 0.75:
        status = "WARNING"

    else:
        status = "NORMAL"

    return {
        **reading,
        "health_score": health_score,
        "status": status
    }


def recursive_analysis(processed, index=0, abnormalities=None):
    """
    Recursively analyzes processed readings until
    the base condition is reached.
    """

    if abnormalities is None:
        abnormalities = []

    # BASE CASE
    if index >= len(processed):
        return abnormalities

    current = processed[index]

    if current["status"] in {"WARNING", "CRITICAL"}:

        abnormalities.append(
            f"Reading {current['id']}: "
            f"{current['status']} "
            f"(health score {current['health_score']}%)"
        )

    # RECURSIVE CALL
    return recursive_analysis(
        processed,
        index + 1,
        abnormalities
    )


def build_report(
    student_inputs,
    generated_count,
    valid,
    invalid,
    processed,
    recursive
):

    abnormal_count = len(recursive)

    if abnormal_count == 0:

        equipment_status = "HEALTHY"

    elif any("CRITICAL" in item for item in recursive):

        equipment_status = "CRITICAL - INSPECTION REQUIRED"

    else:

        equipment_status = "WARNING - MONITOR CLOSELY"

    return {
        "student_specific_inputs": student_inputs,

        "generated_telemetry_data": generated_count,

        "valid_readings": valid,

        "invalid_readings": invalid,

        "processed_results": processed,

        "recursive_analysis": recursive,

        "final_diagnostic_summary": {

            "processed_readings": generated_count,

            "valid_readings": len(valid),

            "invalid_readings": len(invalid),

            "detected_abnormal_conditions": abnormal_count,

            "overall_equipment_status": equipment_status
        }
    }
    from telemetry_data import (
    generate_telemetry,
    validate_reading
)

    from diagnostics import (
    process_reading,
    recursive_analysis,
    build_report
)


# ==========================================================
# STUDENT-SPECIFIC DATA
# ==========================================================

LAST_NAME = "MEDINA"
SEED_NUM = 5
FAVORITE_ARTIST = "ERASERHEAD"


def main():

    print("=" * 70)
    print("INTELLIGENT EQUIPMENT MONITORING PIPELINE")
    print("=" * 70)


    # ======================================================
    # STUDENT-SPECIFIC INPUTS
    # ======================================================

    student_inputs = {
        "LAST_NAME": LAST_NAME,
        "SEED_NUM": SEED_NUM,
        "FAVORITE_ARTIST": FAVORITE_ARTIST
    }


    generated_count = 0
    valid = []
    invalid = []
    processed = []


    print("\nAssessment Data:")

    print(
        "Student-Specific Inputs:",
        student_inputs
    )


    # ======================================================
    # GENERATED TELEMETRY DATA
    # ======================================================

    print("\nGenerated Telemetry Data:")

    # Generator produces one reading at a time.
    for reading in generate_telemetry(
        LAST_NAME,
        SEED_NUM,
        FAVORITE_ARTIST
    ):

        generated_count += 1

        print(reading)


        # ==================================================
        # EXCEPTION HANDLING
        # ==================================================

        try:

            if not validate_reading(reading):

                invalid.append(reading)

                raise ValueError(
                    f"Invalid telemetry at reading "
                    f"{reading['id']}"
                )

            valid.append(reading)

            # Function returns a processed result.
            result = process_reading(reading)

            processed.append(result)


        except (ValueError, TypeError) as error:

            print(
                f"[EXCEPTION HANDLED] {error}"
            )


    # ======================================================
    # RECURSIVE ANALYSIS
    # ======================================================

    recursive = recursive_analysis(processed)


    # ======================================================
    # FINAL REPORT
    # ======================================================

    report = build_report(
        student_inputs,
        generated_count,
        valid,
        invalid,
        processed,
        recursive
    )


    # ======================================================
    # VALID / INVALID RESULTS
    # ======================================================

    print("\nValid/Invalid Results:")

    print(
        f"Valid readings: {len(valid)}"
    )

    print(
        f"Invalid readings: {len(invalid)}"
    )


    # ======================================================
    # PROCESSED RESULTS
    # ======================================================

    print("\nProcessed Results:")

    for item in processed:

        print(
            f"Reading {item['id']}: "
            f"{item['status']} | "
            f"score={item['health_score']}%"
        )


    # ======================================================
    # RECURSIVE ANALYSIS
    # ======================================================

    print("\nRecursive Analysis:")

    if recursive:

        for item in recursive:

            print(item)

    else:

        print(
            "No abnormal conditions detected."
        )


    # ======================================================
    # FINAL DIAGNOSTIC SUMMARY
    # ======================================================

    print("\nFinal Diagnostic Summary:")

    summary = report[
        "final_diagnostic_summary"
    ]

    for key, value in summary.items():

        print(
            f"{key.replace('_', ' ').title()}: "
            f"{value}"
        )


    # ======================================================
    # EXECUTION LOG
    # ======================================================

    print("\nExecution Log:")

    print(
        f"Generator created "
        f"{generated_count} telemetry readings."
    )

    print(
        f"Validated {generated_count} readings "
        f"without terminating on invalid data."
    )

    print(
        f"Processed {len(processed)} valid readings "
        f"using the decorated function."
    )

    print(
        f"Recursive analysis traced "
        f"{len(processed)} processed readings "
        f"to the base case."
    )


    # ======================================================
    # FINAL OUTPUT
    # ======================================================

    print("\nFinal Output:")

    print(
        "Overall Equipment Status:",
        summary["overall_equipment_status"]
    )


# ==========================================================
# PROGRAM ENTRY POINT
# ==========================================================

if __name__ == "__main__":
    main()
```
